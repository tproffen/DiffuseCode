## Powder pattern: X-rays and neutrons

Each figure has a toolbar on its left: the **magnifier** zooms into a box you draw, the
**arrows** pan, **home** goes back to the full view, the **disk** saves a PNG. The readout
in the corner shows the coordinates under the mouse.

Both patterns go into the same axes (`ax=ax`), so they zoom together.

````python
anatase = dc.read_structure("anatase.cif", supercell=(6, 6, 3))

xray = dc.powder(anatase, qmin=0.5, qmax=10, dq=0.005, radiation="xray")
neutron = dc.powder(anatase, qmin=0.5, qmax=10, dq=0.005, radiation="neutron")

ax = xray.plot(label="X-rays")
neutron.plot(ax=ax, label="neutrons")
ax.legend()
ax.set_title("Anatase powder pattern");
````

## Diffuse scattering of a Cu/Au layer

Subtracting the average structure removes the Bragg peaks and leaves the diffuse
scattering from short-range order. Zoom into a region between the Bragg positions to see
its shape; the readout shows h, k and the intensity.

````python
layer = dc.read_structure("cuau_layer.stru")
diffuse = dc.fourier(layer, ll=(0, 0, 0), lr=(4, 0, 0), ul=(0, 4, 0),
                     npoints=(201, 201), subtract_average=100)
ax = diffuse.plot(cmap="magma")
ax.set_aspect("equal")
ax.set_title("Cu/Au layer, diffuse scattering (hk0)");
````

## Pair distribution function of nickel

````python
nickel = dc.read_structure("nickel.cell", supercell=6)
g = dc.pdf(nickel, rmax=20, dr=0.01)
ax = g.plot()
ax.set_title("Nickel, G(r)");
````
